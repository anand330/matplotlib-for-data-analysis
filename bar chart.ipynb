# Matplotlib Bar Chart — Complete Notes

## 1. Bar Chart Kya Hota Hai?

Bar Chart ek type ka graph hai jisme rectangular bars ka use karke different categories ki values compare ki jaati hain.

Har bar ki length ya height us category ki numerical value ko represent karti hai.

**Examples:**
- Different products ki sales compare karna
- Different cities ka revenue compare karna
- Students ke marks compare karna
- Departments mein employees ki count compare karna
- Monthly expenses compare karna

**Data Analysis mein use:** Bar chart se humein easily pata chalta hai ki kaunsi category ki value sabse zyada ya sabse kam hai.

## 2. Vertical Bar Chart — `plt.bar()`

`plt.bar()` ka use vertical bar chart create karne ke liye hota hai.

Ismein:
- X-axis par categories hoti hain.
- Y-axis par numerical values hoti hain.
- Bars vertically upar ki taraf banti hain.

### Syntax

```python
plt.bar(x, height)
```

- `x`: Categories ke names ya positions.
- `height`: Har category ki value.

Example:

```python
products = ["Laptop", "Mobile", "Tablet"]
sales = [50, 80, 35]

plt.bar(products, sales)
plt.show()
```

Is example mein Laptop, Mobile aur Tablet ki sales vertical bars ke through compare hongi.

## 3. Horizontal Bar Chart — `plt.barh()`

`plt.barh()` ka use horizontal bar chart create karne ke liye hota hai.

Ismein:
- Y-axis par categories hoti hain.
- X-axis par numerical values hoti hain.
- Bars left se right ki taraf banti hain.

### Syntax

```python
plt.barh(y, width)
```

- `y`: Categories ke names ya positions.
- `width`: Har category ki numerical value.

Example:

```python
plt.barh(products, sales)
plt.show()
```

### `plt.bar()` aur `plt.barh()` mein Difference

- `plt.bar()` → Vertical bars.
- `plt.barh()` → Horizontal bars.

Jab category names lambe hon, tab horizontal bar chart aksar zyada readable hota hai.

## 4. Title, X-axis aur Y-axis

Chart ko understandable banane ke liye title aur axis labels use karte hain.

```python
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Units Sold")
```

Vertical chart mein categories X-axis par aur values Y-axis par hoti hain.

Horizontal chart mein categories Y-axis par aur values X-axis par hoti hain.

## 5. Bar Colors — `color`

`color` parameter bars ka color set karta hai.

```python
plt.bar(products, sales, color="green")
```

Har bar ka color alag bhi rakh sakte hain:

```python
plt.bar(products, sales, color=["red", "blue", "green"])
```

Colors ka use meaningful comparison ke liye karo; unnecessary colors chart ko confusing bana sakte hain.

## 6. Bar Width — `width`

`width` parameter vertical bars ki width control karta hai.

```python
plt.bar(products, sales, width=0.5)
```

Vertical bar chart mein default width `0.8` hoti hai.

Horizontal bar chart mein bar ki thickness control karne ke liye `height` parameter use hota hai.

```python
plt.barh(products, sales, height=0.5)
```

## 7. Bar Borders — `edgecolor` and `linewidth`

Bars ki boundary ka color aur thickness set kar sakte hain.

```python
plt.bar(
    products,
    sales,
    edgecolor="black",
    linewidth=1.5
)
```

- `edgecolor`: Bar ki border ka color.
- `linewidth`: Border ki thickness.

## 8. Multiple Categories Compare Karna

Bar chart ka sabse common use multiple categories ki values compare karna hai.

Example: Different cities ki sales.

```python
cities = ["Delhi", "Mumbai", "Jaipur", "Pune"]
sales = [50000, 70000, 40000, 60000]

plt.bar(cities, sales)
```

Isse easily compare kar sakte hain ki kis city ki sales sabse zyada hai.

## 9. Multiple Datasets — Grouped Bar Chart

Jab har category ke liye do datasets compare karne hon, jaise 2025 aur 2026 ki sales, tab grouped bar chart useful hota hai.

Grouped chart mein bars ko side-by-side place kiya jata hai.

Iske liye NumPy ka use karke bar positions calculate kar sakte hain.

```python
import numpy as np

x = np.arange(len(products))
width = 0.35

plt.bar(x - width/2, sales_2025, width, label="2025")
plt.bar(x + width/2, sales_2026, width, label="2026")

plt.xticks(x, products)
plt.legend()
```

Yahan:
- `np.arange()` category positions banata hai.
- `width` har bar ki width define karta hai.
- `x - width/2` aur `x + width/2` bars ko side-by-side place karte hain.
- `plt.xticks()` positions par category names show karta hai.

## 10. Stacked Bar Chart

Stacked bar chart mein multiple datasets ki bars ek doosre ke upar stack hoti hain.

Iska use category ke total aur uske parts compare karne ke liye hota hai.

```python
plt.bar(products, online_sales, label="Online")
plt.bar(products, store_sales, bottom=online_sales, label="Store")

plt.legend()
```

`bottom=online_sales` second dataset ki bars ko pehle dataset ke upar place karta hai.

## 11. Bar Values Show Karna — `plt.bar_label()`

Bars par exact numerical values display karne ke liye `plt.bar_label()` use hota hai.

```python
bars = plt.bar(products, sales)
plt.bar_label(bars)
```

Ye values ko chart se read karna easy banata hai.

## 12. Grid Lines — `plt.grid()`

Grid lines numerical values compare karne mein help karti hain.

Vertical bar chart mein Y-axis ki grid lines useful hoti hain:

```python
plt.grid(axis="y", linestyle="--", alpha=0.5)
```

Horizontal bar chart mein X-axis ki grid lines use kar sakte hain:

```python
plt.grid(axis="x", linestyle="--", alpha=0.5)
```

## 13. Figure Size — `plt.figure()`

`figsize` chart ki width aur height inches mein set karta hai.

```python
plt.figure(figsize=(8, 5))
```

Zyada categories hon to figure width increase karna useful hota hai.

## 14. Axis Limits and Ticks

Axis limits chart mein visible numerical range control karti hain.

Vertical chart:

```python
plt.ylim(0, 100)
```

Horizontal chart:

```python
plt.xlim(0, 100)
```

Ticks customize karne ke liye:

```python
plt.yticks([0, 20, 40, 60, 80, 100])
```

Limits aur ticks ko data ke according set karna chahiye. Aisi range mat choose karo jo important values ko hide ya misleading bana de.

## 15. Category Labels Rotate Karna

Agar category names lambe hain aur overlap ho rahe hain, to labels rotate kar sakte hain.

```python
plt.xticks(rotation=45)
```

Horizontal bar chart long category names ke liye ek achha alternative ho sakta hai.

## 16. Bar Chart Mein Legend

Jab ek chart mein multiple datasets hon, tab legend use karte hain.

```python
plt.bar(products, sales_2025, label="2025")
plt.bar(products, sales_2026, label="2026")
plt.legend()
```

Single dataset ke simple bar chart mein legend zaroori nahi hota.

## 17. `plt.tight_layout()`

`plt.tight_layout()` labels, title aur chart ke beech spacing adjust karta hai.

```python
plt.tight_layout()
```

Ye long category names ya rotated labels ke saath useful hota hai.

## 18. Complete Bar Chart Example

```python
import matplotlib.pyplot as plt

sales = [50, 80, 35, 60]
```

````python

#Grouped Bar Chart — 2025 vs 2026
import numpy as np
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Headphones"]
sales_2025 = [50, 80, 35, 60]
sales_2026 = [65, 90, 45, 75]

x = np.arange(len(products))
width = 0.35

plt.bar(x - width / 2, sales_2025, width, label="2025")
plt.bar(x + width / 2, sales_2026, width, label="2026")

plt.xticks(x, products)
plt.title("Product Sales: 2025 vs 2026")
plt.xlabel("Products")
plt.ylabel("Units Sold")
plt.legend()

plt.show()
````

````python
# Stacked Bar Chart
products =["laptop","mobile","tablet","haedphones"]
online_sales =[30,50,20,40]
store_sales=[20,30,15,20]

plt.bar(products, online_sales , label="online_sales")
plt.bar(products , store_sales , bottom=online_sales , label="store sales")

plt.title("online vs store sales")
plt.xlabel("products")
plt.ylabel("units sold")
plt.legend()

plt.show()
````

# Matplotlib Bar Chart — Interview Notes

## 1. What is a Bar Chart?
A bar chart uses rectangular bars to compare numerical values across categories.

## 2. Which function creates a vertical bar chart?
`plt.bar(x, height)`

## 3. Which function creates a horizontal bar chart?
`plt.barh(y, width)`

## 4. Difference between `plt.bar()` and `plt.barh()`
- `plt.bar()` → Vertical bars.
- `plt.barh()` → Horizontal bars.

## 5. What is the `width` parameter?
It controls the width of vertical bars.

## 6. What is the `height` parameter in `plt.barh()`?
It controls the thickness of horizontal bars.

## 7. How do you show values on bars?
```python
bars = plt.bar(x, height)
plt.bar_label(bars)
```

## 8. What is a Grouped Bar Chart?
A chart that places bars side-by-side to compare multiple datasets across the same categories.

## 9. What is a Stacked Bar Chart?
A chart in which multiple datasets are stacked on top of one another to show totals and their components.

## 10. What is the purpose of `bottom`?
It specifies the vertical starting position of bars, commonly used to create stacked bar charts.

## 11. What is `edgecolor`?
It sets the border color of bars.

## 12. How do you add grid lines?
```python
plt.grid(axis="y")
```

For a horizontal bar chart:
```python
plt.grid(axis="x")
```

## 13. When should you use a horizontal bar chart?
When category names are long or there are many categories.

## 14. When should you use a bar chart?
When comparing values across discrete categories, such as products, cities or departments.

## 15. Important Parameters
- `color` — Bar color
- `width` — Vertical bar width
- `height` — Horizontal bar thickness
- `label` — Dataset label
- `bottom` — Stacking position
- `edgecolor` — Border color
- `linewidth` — Border thickness
- `plt.bar_label()` — Values on bars
- `plt.legend()` — Legend
- `plt.tight_layout()` — Layout spacing

````python
import pandas as pd

df = pd.DataFrame({
    "Product": [
        "Laptop", "Mobile", "Tablet", "Keyboard", "Mouse",
        "Monitor", "Printer", "Headphones"
    ],
    "Sales": [85000, 120000, 45000, 18000, 12000, 55000, 30000, 22000],
    "Profit": [12000, 20000, 7000, 4000, 3000, 9000, 5000, 6000],
    "Units_Sold": [10, 25, 15, 40, 60, 18, 12, 30],
    "Marketing_Cost": [5000, 8000, 3000, 1500, 1000, 3500, 2000, 2500]
})
df
````

````python
df
bars=plt.bar(df["Product"] , df["Sales"], color="blue" , edgecolor="black", linewidth=2, label="product wise sales", alpha=0.5)
plt.title("product wise sales")
plt.grid(axis="y",linestyle="--",alpha=0.9)
plt.xlabel("product")
plt.ylabel("sales")
plt.legend()
plt.bar_label(bars)
plt.show()
````
