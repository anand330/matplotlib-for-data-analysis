# Matplotlib Histogram

## 1. Histogram Kya Hai?
Histogram numerical data ki distribution show karta hai. Isse pata chalta hai ki data ki values kis range mein kitni baar aati hain.

**Example:** Students ke marks, employees ki salary, customers ki age.

## 2. `plt.hist()`
Histogram create karne ke liye use hota hai.

```python
plt.hist(data)
```

## 3. Bins
Bins numerical data ke intervals hote hain. `bins=5` ka matlab hai data ko 5 intervals mein divide karna.

```python
plt.hist(data, bins=5)
```

Custom intervals bhi de sakte hain:

```python
plt.hist(data, bins=[0, 20, 40, 60, 80, 100])
```

## 4. Frequency
Frequency batati hai ki kisi bin ke andar kitni observations hain. Default histogram mein Y-axis frequency show karti hai.

```python
plt.ylabel("Frequency")
```

## 5. Distribution
Distribution batati hai ki data ki values kaise spread hui hain.

- **Symmetric:** Dono sides ka shape similar.
- **Right-skewed:** Right side ki tail lambi.
- **Left-skewed:** Left side ki tail lambi.
- **Uniform:** Intervals mein frequency roughly similar.
- **Bimodal:** Do prominent peaks.

## 6. Histogram vs Bar Chart
- `plt.hist()` numerical data ki distribution show karta hai.
- `plt.bar()` categories ki values compare karta hai.
- Histogram ke bars usually touch karte hain; bar chart mein usually gaps hote hain.

## 7. Color, Border and Transparency
`color` bar ka color, `edgecolor` border ka color aur `alpha` transparency control karta hai.

```python
plt.hist(data, bins=5, color="skyblue",
         edgecolor="black", alpha=0.7)
```

`alpha=1` opaque aur `alpha=0` fully transparent hota hai.

## 8. Relative Frequency
Relative frequency kisi bin ki frequency ko total observations se divide karke milti hai.

```python
weights = np.ones(len(data)) / len(data)
plt.hist(data, bins=5, weights=weights)
```

## 9. Density Histogram
`density=True` histogram ko normalize karta hai, jisse total area 1 hota hai. Y-axis frequency ke bajaye probability density show karti hai.

```python
plt.hist(data, bins=5, density=True)
```

## 10. Cumulative Histogram
Cumulative histogram har bin tak ki accumulated frequency show karta hai.

```python
plt.hist(data, bins=5, cumulative=True)
```

## 11. Histogram Interpretation
Histogram dekhte waqt ye points check karo:

- **Center:** Data kis range ke aas-paas hai?
- **Spread:** Data kitna phaila hua hai?
- **Shape:** Symmetric hai ya skewed?
- **Peaks:** Ek peak hai ya multiple?
- **Gaps:** Kya kuch ranges mein data missing hai?
- **Possible outliers:** Kya kuch values baaki data se kaafi door hain?

## 12. Important Chart Settings

```python
plt.title("Data Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.grid(axis="y", alpha=0.4)
plt.show()
```

- `plt.title()` chart ka title set karta hai.
- `plt.xlabel()` aur `plt.ylabel()` axis labels set karte hain.
- `plt.grid()` grid lines add karta hai.
- `plt.show()` chart display karta hai.

## 13. Important Parameters

| Parameter | Use |
|---|---|
| `bins` | Number of bins ya custom bin edges |
| `color` | Bar color |
| `edgecolor` | Border color |
| `alpha` | Transparency |
| `density=True` | Density histogram |
| `cumulative=True` | Cumulative histogram |
| `weights` | Observations ko weights dena |

````python
import matplotlib.pyplot as plt
# Basic Histogram
marks=[
    45,50, 55, 60 , 65, 32, 45 ,76, 67,12,34,45, 32,89,56]
plt.hist(marks)
plt.show()
````

````python
# Bins Customize Karna
plt.hist(marks,bins=[10,20,30,40,50 ,60 ,70 ,80,90] , edgecolor="black")
plt.show()
````

````python
# Custom Bin Edges
plt.hist(
    marks, bins=5 , edgecolor="black")
plt.title("marks by score range")
plt.xlabel("marks range")
plt.ylabel("frequency")
````

````python
# Colors, Transparency and Grid
plt.figure(figsize=(10,8))
plt.hist(marks,bins=6,
         edgecolor="black",
         color="green",
         alpha=0.9
        )
plt.title("distribution of student marks")
plt.xlabel("marks")
plt.ylabel("frequency")
plt.grid(axis="y", linestyle="--",alpha=0.9)
plt.show()
````

````python
import numpy as np

weights = np.ones(len(marks)) / len(marks)

plt.hist(
    marks,
    bins=5,
    weights=weights,
    edgecolor="black"
)

plt.title("Relative Frequency of Marks")
plt.xlabel("Marks")
plt.ylabel("Proportion")

plt.show()
````

````python
# Density Histogram
plt.hist(
    marks,
    bins=5,
    density=True,
    color="orange",
    edgecolor="black"
)

plt.title("Probability Density of Marks")
plt.xlabel("Marks")
plt.ylabel("Density")

plt.show()
````

````python
# Cumulative Histogram
plt.hist(
    marks,
    bins=[10,20,30,40,50,60,70,80,90],
    cumulative=True,
    color="green",
    edgecolor="black"
)

plt.title("Cumulative Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Cumulative Frequency")

plt.show()
````

````python
group_a = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75]
group_b = [60, 65, 70, 72, 75, 78, 80, 85, 88, 92]

common_bins = [40, 50, 60, 70, 80, 90, 100]

plt.hist(
    group_a,
    bins=common_bins,
    alpha=0.6,
    edgecolor="black",
    label="Group A"
)

plt.hist(
    group_b,
    bins=common_bins,
    alpha=0.6,
    edgecolor="black",
    label="Group B"
)

plt.title("Marks Distribution Comparison")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.legend()

plt.show()
````

````python
salary = [
    18000, 20000, 22000, 25000, 27000,
    30000, 32000, 35000, 38000, 40000,
    45000, 50000, 55000, 60000, 90000
]
plt.figure(figsize=(8,5))

plt.hist(salary,
         bins=[10000, 20000 , 30000 ,40000 ,50000 , 60000 , 70000 , 80000, 90000]
         ,
         color="green",
         edgecolor="black",
         label="salary",
         alpha=0.9)
plt.title("employee salary distibution")
plt.xlabel("salary")
plt.ylabel("frequency")
plt.grid(axis="y" , linestyle="--",alpha=0.9)
plt.legend()
plt.show()
````

# Histogram — Interview Revision

**1. What is a histogram?**  
It shows the distribution of numerical data using intervals.

**2. Which function creates a histogram?**  
`plt.hist(data)`

**3. What are bins?**  
Bins are intervals used to group numerical values.

**4. What is frequency?**  
The number of observations in a bin.

**5. Histogram vs Bar Chart?**
- Histogram: Numerical data distribution.
- Bar chart: Category comparison.

**6. What is a right-skewed distribution?**  
A distribution with a longer tail towards higher values.

**7. What is a left-skewed distribution?**  
A distribution with a longer tail towards lower values.

**8. What does `density=True` do?**  
Normalizes the histogram so its total area equals 1.

**9. What does `cumulative=True` do?**  
Shows accumulated frequency across bins.

**10. What should you check when interpreting a histogram?**
- Center
- Spread
- Shape
- Peaks
- Gaps
- Possible outliers

**11. What does `alpha` control?**  
Bar transparency.

**12. Why does bin selection matter?**  
Too few bins can hide patterns; too many can make random variation look important.
