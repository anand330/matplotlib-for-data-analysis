````python
import matplotlib.pyplot as plt
months=["jan","feb","mar","apr","may","jun","july"]
sales=[120,150,130,180,220,250,400]
profit=[30,40,35,50,65,70,80]
plt.figure(figsize=(10,10))


plt.plot(months,
         sales,
         marker="o",
         markersize=5,
         markerfacecolor="red",
         markeredgecolor="black",
         color="green",
         linestyle="--",
         linewidth=2,
         label="sales"
        )
plt.plot(
    months,
    profit,
    color="red",
    marker="s",
    linestyle="--",
    linewidth=2,
    markersize=5,
    markerfacecolor="red",
    markeredgecolor="black",
    label="profit"
)
plt.title("monthly sales and profit")
plt.xlabel("months")
plt.ylabel("amount")
plt.grid(axis="x",linestyle=":",alpha=0.5)
plt.grid(axis="y", linestyle=":",alpha=0.9)
plt.legend()
plt.tight_layout()
plt.show()
````

# Matplotlib Line Chart — Complete Notes

## 1. Line Chart Kya Hota Hai?

Line Chart ek type ka graph hai jisme data points ko lines se connect kiya jata hai. Iska use data mein changes aur trends ko time ke saath samajhne ke liye hota hai.

**Examples:**
- Month-wise sales analysis
- Daily temperature analysis
- Company ke profit ka trend
- Website visitors ka analysis
- Stock prices mein changes

**Data Analysis mein use:** Line chart humein batata hai ki data increase ho raha hai, decrease ho raha hai ya stable hai.

## 2. Matplotlib Import Karna

Matplotlib Python ki ek visualization library hai. Iska use graphs aur charts create karne ke liye hota hai.

```python
import matplotlib.pyplot as plt
```

Yahan:
- `matplotlib` ek visualization library hai.
- `pyplot` graph banane ke liye functions provide karta hai.
- `plt` iska short alias hai.

## 3. Basic Line Chart — `plt.plot()`

`plt.plot()` ka use line chart create karne ke liye hota hai.

### Syntax

```python
plt.plot(x, y)
```

- `x`: X-axis par aane wali values.
- `y`: Y-axis par aane wali values.

Dono lists mein values ki quantity same honi chahiye.

Example:

```python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 200]

plt.plot(months, sales)
plt.show()
```

Is example mein months X-axis par aur sales Y-axis par show hongi.

## 4. X-axis aur Y-axis

### X-axis — `plt.xlabel()`

X-axis horizontal axis hoti hai. Is par generally months, dates, categories ya independent variable show hota hai.

```python
plt.xlabel("Months")
```

### Y-axis — `plt.ylabel()`

Y-axis vertical axis hoti hai. Is par generally sales, profit, marks, temperature ya measured values show hoti hain.

```python
plt.ylabel("Sales")
```

### Chart Title — `plt.title()`

Chart ka title graph ka purpose batata hai.

```python
plt.title("Monthly Sales")
```

## 5. Chart Display Karna — `plt.show()`

`plt.show()` chart ko display karta hai.

```python
plt.plot(months, sales)
plt.show()
```

Jupyter Notebook mein simple plots kabhi-kabhi automatically bhi display hote hain, lekin `plt.show()` use karna achhi practice hai.

## 6. Multiple Lines

Jab ek hi chart mein do ya do se zyada datasets compare karne hon, tab multiple lines use karte hain.

Example: Sales aur Profit ko compare karna.

```python
plt.plot(months, sales, label="Sales")
plt.plot(months, profit, label="Profit")
plt.legend()
```

- Har `plt.plot()` call ek line create karta hai.
- `label` line ka naam define karta hai.
- `plt.legend()` chart par line names display karta hai.

Multiple lines se comparison easy ho jata hai.

## 7. Markers

Markers line chart ke data points ko represent karte hain. Inse pata chalta hai ki actual data points kahan hain.

Important markers:

| Marker | Meaning |
|---|---|
| `'o'` | Circle |
| `'.'` | Point |
| `'*'` | Star |
| `'s'` | Square |
| `'^'` | Upward triangle |
| `'D'` | Diamond |
| `'x'` | Cross |

Example:

```python
plt.plot(months, sales, marker="o")
```

### Marker Size — `markersize`

Marker ka size change karne ke liye:

```python
plt.plot(months, sales, marker="o", markersize=8)
```

### Marker Face Color — `markerfacecolor`

Marker ke andar ka color change karta hai.

```python
plt.plot(months, sales, marker="o", markerfacecolor="yellow")
```

### Marker Edge Color — `markeredgecolor`

Marker ki boundary ka color change karta hai.

```python
plt.plot(months, sales, marker="o", markeredgecolor="red")
```

## 8. Line Styles

`linestyle` parameter line ka pattern change karta hai.

| Style | Meaning |
|---|---|
| `'-'` | Solid line |
| `'--'` | Dashed line |
| `':'` | Dotted line |
| `'-.'` | Dash-dot line |

Example:

```python
plt.plot(months, sales, linestyle="--")
```

Line ki thickness change karne ke liye `linewidth` use hota hai.

```python
plt.plot(months, sales, linewidth=3)
```

## 9. Line Colors

`color` parameter line ka color set karta hai.

```python
plt.plot(months, sales, color="red")
```

Common colors: `"red"`, `"blue"`, `"green"`, `"black"`, `"orange"`, `"purple"`.

Color, marker aur line style ko ek saath bhi set kar sakte hain.

```python
plt.plot(
    months,
    sales,
    color="green",
    linestyle="--",
    linewidth=2,
    marker="o",
    markersize=7
)
```

## 10. Legend — `plt.legend()`

Legend batata hai ki chart mein kaunsi line kis dataset ko represent kar rahi hai.

```python
plt.plot(months, sales, label="Sales")
plt.plot(months, profit, label="Profit")
plt.legend()
```

Legend ke bina multiple lines ko identify karna difficult ho sakta hai.

## 11. Grid Lines — `plt.grid()`

Grid lines graph ko read karna aur values compare karna easy banati hain.

```python
plt.grid()
```

Sirf horizontal grid lines:

```python
plt.grid(axis="y")
```

Grid ki style aur transparency bhi set kar sakte hain.

```python
plt.grid(axis="y", linestyle="--", alpha=0.5)
```

`alpha` transparency control karta hai. Iski value `0` se `1` ke beech hoti hai.

## 12. Axis Limits

Axis limits decide karti hain ki chart par values ki kaunsi range show hogi.

### X-axis limits — `plt.xlim()`

```python
plt.xlim(0, 5)
```

### Y-axis limits — `plt.ylim()`

```python
plt.ylim(0, 300)
```

Limits set karte waqt ensure karo ki important data points graph se bahar na ho jayein.

## 13. Axis Ticks

Ticks axis par show hone wali positions ya values hoti hain.

### X-axis ticks — `plt.xticks()`

```python
plt.xticks(rotation=45)
```

Isse X-axis ke labels 45 degrees rotate honge.

Custom tick positions aur labels bhi set kar sakte hain.

```python
plt.xticks([0, 1, 2, 3], ["Jan", "Feb", "Mar", "Apr"])
```

### Y-axis ticks — `plt.yticks()`

```python
plt.yticks([0, 50, 100, 150, 200, 250])
```

Ticks ko data aur chart ke scale ke according choose karna chahiye.

## 14. Figure Size — `plt.figure()`

`plt.figure(figsize=(width, height))` chart ki size set karta hai. Size inches mein specify hoti hai.

```python
plt.figure(figsize=(8, 5))
```

Isse chart 8 inches wide aur 5 inches high hoga.

## 15. Multiple Lines Ko Customize Karna

Har line ka color, marker, line style aur label alag ho sakta hai.

```python
plt.plot(months, sales, color="blue", marker="o",
         linestyle="-", label="Sales")

plt.plot(months, profit, color="red", marker="s",
         linestyle="--", label="Profit")
```

Isse dono datasets ko visually compare karna easy hota hai.

## 16. `plt.tight_layout()`

`plt.tight_layout()` chart ke title aur labels ke beech spacing ko adjust karne mein help karta hai, taaki labels cut ya overlap na hon.

```python
plt.tight_layout()
```


# Matplotlib Line Chart — Interview Revision

## Important Questions and Answers

**1. What is a line chart?**  
A line chart connects data points using lines to show trends and changes.

**2. Which function creates a line chart?**
```python
plt.plot(x, y)
```

**3. What is the difference between X-axis and Y-axis?**
- X-axis: Independent variable or categories.
- Y-axis: Values being measured.

**4. How do you create multiple lines?**  
Use multiple `plt.plot()` calls.

**5. What is a marker?**  
A marker represents an individual data point.

**6. Name important marker symbols.**
- `'o'` — Circle
- `'s'` — Square
- `'^'` — Triangle
- `'*'` — Star
- `'D'` — Diamond

**7. What is `linestyle`?**  
It controls the pattern of a line.
- `'-'` — Solid
- `'--'` — Dashed
- `':'` — Dotted
- `'-.'` — Dash-dot

**8. What is the difference between `markersize` and `linewidth`?**
- `markersize`: Controls marker size.
- `linewidth`: Controls line thickness.

**9. What is `plt.legend()`?**  
It displays labels that identify different lines.

**10. What is `plt.grid()`?**  
It adds grid lines to help read and compare values.

**11. What is the purpose of `plt.xlim()` and `plt.ylim()`?**  
They set the visible range of the X-axis and Y-axis.

**12. What does `plt.figure(figsize=(8, 5))` do?**  
It sets the figure size to 8 inches wide and 5 inches high.

**13. What is `plt.tight_layout()`?**  
It adjusts layout spacing to reduce overlap and prevent labels from being cut off.

**14. Why is a line chart useful in Data Analysis?**  
It helps identify trends, increases, decreases and changes over time.
