# Matplotlib Fundamentals

## 1. What is Matplotlib?

Matplotlib is a Python library used to create graphs, charts, and visualizations.

It helps us understand data visually instead of looking only at numbers.

**Examples:**
- Line chart
- Bar chart
- Scatter plot
- Histogram
- Pie chart
- Box plot

## 2. Uses of Matplotlib in Data Analysis

Matplotlib is used to:

- Compare values between categories.
- Analyze trends over time.
- Understand the distribution of data.
- Identify unusual values and possible outliers.
- Visualize relationships between variables.
- Present data analysis results clearly.

**Example:** A Data Analyst can use Matplotlib to visualize monthly sales and identify which month had the highest sales.

## 3. Import Matplotlib

The `pyplot` module provides functions for creating and customizing charts.

### Syntax
```python
import matplotlib.pyplot as plt
```

Here:
- `matplotlib` is the visualization library.
- `pyplot` provides plotting functions.
- `plt` is the commonly used alias for `matplotlib.pyplot`.

## 4. What is a Figure?

A **Figure** is the complete canvas or container on which a chart is created.

It can contain one or more Axes.

### Syntax
```python
fig = plt.figure()
```

### Set Figure Size
```python
fig = plt.figure(figsize=(8, 5))
```

- `figsize` specifies the figure width and height in inches.
- `plt.figure()` creates a new figure.

## 5. What are Axes?

An **Axes** is the area inside a Figure where data is plotted. It usually contains the X-axis, Y-axis, labels, title, and plotted data.

### Syntax
```python
fig, ax = plt.subplots()
```

Here:
- `fig` represents the Figure.
- `ax` represents the Axes object.

You can use `ax` to create and customize a chart.

### Example
```python
fig, ax = plt.subplots()

ax.plot([1, 2, 3], [10, 20, 30])
ax.set_title("Sample Line Chart")
ax.set_xlabel("X Values")
ax.set_ylabel("Y Values")

plt.show()
```

## 6. Difference Between Figure and Axes

| Figure | Axes |
|---|---|
| Complete canvas | Area where data is plotted |
| Can contain multiple Axes | Contains the plotted chart |
| Created using `plt.figure()` or `plt.subplots()` | Created using `plt.subplots()` or other Axes methods |

**Remember:** One Figure can contain multiple Axes, allowing multiple charts in the same figure.

## 7. What is `plt.show()`?

`plt.show()` displays the created chart.

It is commonly placed at the end of the plotting code.

### Syntax
```python
plt.show()
```

In Jupyter Notebook, charts may appear automatically after a plotting cell runs. However, using `plt.show()` makes the intention to display the chart explicit and is useful in scripts.

## 8. Basic Matplotlib Workflow

1. Import Matplotlib.
2. Prepare the data.
3. Create a Figure and Axes, if needed.
4. Create the chart.
5. Add a title and axis labels.
6. Display the chart using `plt.show()`.

## 9. Complete Example

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 180, 220]

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(months, sales, marker="o")

ax.set_title("Monthly Sales")
ax.set_xlabel("Months")
ax.set_ylabel("Sales")

plt.show()
```

This example creates a Figure, adds an Axes, plots monthly sales, labels the chart, and displays it.

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 130, 200]
profit = [20, 35, 30, 50]

axes[0].plot(months, sales)
axes[0].set_title("Monthly Sales")

axes[1].plot(months, profit)
axes[1].set_title("Monthly Profit")

plt.show()
````

````python
fig, ax = plt.subplots(figsize=(10,5))

months=["jan","feb","mar","apr","may","junu","july","aug","sep","oct"]
sales_2025=[20000,30000,40000,50000,40000,30000,70000,90000,100000,10000]
sales_2026=[50000,60000,80000,90000,10000,100000,20000,30000,40000,10000]
sales_2027=[100000,200000,300000,10000,40000,50000,60000,100000,40000,60000]

plt.plot(months,sales_2025,label="sales_2025",
        marker="s",
        markersize=5,
        markerfacecolor="blue",
        markeredgecolor="black",
        color="red",
        linestyle="--",
        linewidth=2)

plt.plot(
    months,
    sales_2026,
    marker="o",
    markersize=5,
    markerfacecolor="red",
    markeredgecolor="black",
    linestyle="--",
    linewidth=2,
    color="blue"
)

plt.plot(
    months , sales_2027,
    color="green",
    linestyle="--",
    linewidth=2,
    marker="s",
    markersize=5,
    markerfacecolor="green",
    markeredgecolor="black"
)
plt.show()
````

# Matplotlib Fundamentals — Interview Notes

**1. What is Matplotlib?**  
A Python library used to create charts and visualizations.

**2. Why is Matplotlib used in Data Analysis?**  
To visualize trends, compare categories, study distributions, and identify patterns.

**3. How do you import Matplotlib?**
```python
import matplotlib.pyplot as plt
```

**4. What is a Figure?**  
The complete canvas that contains one or more Axes.

**5. What are Axes?**  
The plotting area where data, axes, titles, labels, and other chart elements appear.

**6. What is `plt.show()`?**  
It displays the generated chart.

**7. What is `figsize`?**  
It sets the Figure dimensions in inches.

**8. What is the difference between Figure and Axes?**  
Figure is the complete canvas; Axes is the area where data is plotted.

**9. What does `plt.subplots()` return?**  
It returns a Figure object and one or more Axes objects.

**10. Difference between `plt.plot()` and `ax.plot()`?**
- `plt.plot()` uses Matplotlib's state-based plotting interface.
- `ax.plot()` plots directly on a specific Axes object.

**Important syntax**
```python
fig = plt.figure()
fig, ax = plt.subplots()
plt.show()
```
